```python
!pip install openai -q

import openai
import time
import os

os.environ["OPENAI_API_KEY"] = "***REMOVED-SECRET***"

client = openai.OpenAI()

def call_openai(prompt, temperature=0.7):
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}],
        temperature=temperature
    )
    return response.choices[0].message.content.strip()

def run_multiple(prompt, n=5):
    results = []
    for i in range(n):
        result = call_openai(prompt)
        results.append(result)
        print(f"Run {i+1}: {result}")
        time.sleep(0.5)
    return results

print("✅ Tudo pronto!")
```

```python
# VERSÃO 1 — Prompts simples
sentiment_v1 = """
Classify this customer message: "I love this product! It's exactly what I needed."
"""

product_v1 = """
Create a product description for a wireless mouse that costs $29.99.
"""

extraction_v1 = """
Extract information from this customer feedback: "I ordered item #12345 on March 15th. The delivery was fast but the packaging was damaged."
"""

print("=== SENTIMENTO ===")
print(call_openai(sentiment_v1))

print("\n=== PRODUTO ===")
print(call_openai(product_v1))

print("\n=== EXTRAÇÃO ===")
print(call_openai(extraction_v1))
```

```python
print("--- SENTIMENT (5 runs) ---")
sentiment_5runs = run_multiple(sentiment_v1, n=5)

print("\n--- PRODUCT (5 runs) ---")
product_5runs = run_multiple(product_v1, n=5)

print("\n--- EXTRACTION (5 runs) ---")
extraction_5runs = run_multiple(extraction_v1, n=5)
```

```python
print("--- SENTIMENT (10 runs) ---")
sentiment_10runs = run_multiple(sentiment_v1, n=10)

print("\n--- PRODUCT (10 runs) ---")
product_10runs = run_multiple(product_v1, n=10)

print("\n--- EXTRACTION (10 runs) ---")
extraction_10runs = run_multiple(extraction_v1, n=10)
```

```python
print("--- SENTIMENT (15 runs) ---")
sentiment_15runs = run_multiple(sentiment_v1, n=15)

print("\n--- PRODUCT (15 runs) ---")
product_15runs = run_multiple(product_v1, n=15)

print("\n--- EXTRACTION (15 runs) ---")
extraction_15runs = run_multiple(extraction_v1, n=15)
```

```python
def count_unique(results):
    return len(set(results))

def consistency_score(results):
    most_common = max(set(results), key=results.count)
    return round((results.count(most_common) / len(results)) * 100, 1)

print("SENTIMENT V1:", consistency_score(sentiment_15runs), "%")
print("PRODUCT V1:", consistency_score(product_15runs), "%")
print("EXTRACTION V1:", consistency_score(extraction_15runs), "%")
```

```python
sentiment_v2 = """You are a sentiment classifier. Respond with ONLY one word: Positive, Negative, or Neutral.

Customer message: "I love this product! It's exactly what I needed."

Sentiment:"""

product_v2 = """You are an e-commerce copywriter. Use this exact structure:

**Product Name:** [name]
**Price:** [price]
**Description:** [2-3 sentences]
**Key Features:**
- [feature 1]
- [feature 2]
- [feature 3]

Product: Wireless mouse, $29.99"""

extraction_v2 = """Extract information and return in this exact format:

Order Number: [value or N/A]
Order Date: [value or N/A]
Delivery Speed: [value or N/A]
Packaging Condition: [value or N/A]

Text: "I ordered item #12345 on March 15th. The delivery was fast but the packaging was damaged."

Extracted Data:"""

print("=== SENTIMENT V2 ===")
print(call_openai(sentiment_v2))

print("\n=== PRODUCT V2 ===")
print(call_openai(product_v2))

print("\n=== EXTRACTION V2 ===")
print(call_openai(extraction_v2))
```

```python
sentiment_v2_15 = run_multiple(sentiment_v2, n=15)
product_v2_15 = run_multiple(product_v2, n=15)
extraction_v2_15 = run_multiple(extraction_v2, n=15)

print("SENTIMENT V2:", consistency_score(sentiment_v2_15), "%")
print("PRODUCT V2:", consistency_score(product_v2_15), "%")
print("EXTRACTION V2:", consistency_score(extraction_v2_15), "%")
```

```python
sentiment_v3 = """You are a sentiment classifier. Respond with ONLY one word: Positive, Negative, or Neutral.

Examples:
Message: "This product is amazing, I love it!"
Sentiment: Positive

Message: "Terrible quality, broke after one day."
Sentiment: Negative

Message: "The package arrived on Tuesday."
Sentiment: Neutral

Message: "It works fine, nothing special."
Sentiment: Neutral

Now classify this:
Message: "I love this product! It's exactly what I needed."
Sentiment:"""

product_v3 = """You are an expert e-commerce copywriter. Always use this exact format:

EXAMPLE:
Product: Bluetooth speaker, $49.99
**Product Name:** SoundWave Bluetooth Speaker
**Price:** $49.99
**Description:** Powerful portable speaker with 360-degree sound. Perfect for outdoor adventures or home use. Features 12-hour battery life.
**Key Features:**
- 360-degree immersive sound
- 12-hour battery life
- Waterproof design
- Bluetooth 5.0 connectivity
- Built-in microphone

Now write for this product:
Product: Wireless mouse, $29.99"""

extraction_v3 = """You are a data extraction specialist. Follow these steps:

Step 1: Read the text carefully
Step 2: Find the order number (digits after #)
Step 3: Find the date mentioned
Step 4: Find delivery speed (fast/slow/normal)
Step 5: Find packaging condition (good/damaged/broken)
Step 6: Return ONLY this JSON

Text: "I ordered item #12345 on March 15th. The delivery was fast but the packaging was damaged."

{
  "order_number": "12345",
  "order_date": "March 15th",
  "delivery_speed": "fast",
  "packaging_condition": "damaged"
}"""

print("=== SENTIMENT V3 ===")
print(call_openai(sentiment_v3))

print("\n=== PRODUCT V3 ===")
print(call_openai(product_v3))

print("\n=== EXTRACTION V3 ===")
print(call_openai(extraction_v3))
```

```python
sentiment_v3_15 = run_multiple(sentiment_v3, n=15)
product_v3_15 = run_multiple(product_v3, n=15)
extraction_v3_15 = run_multiple(extraction_v3, n=15)

print("SENTIMENT V3:", consistency_score(sentiment_v3_15), "%")
print("PRODUCT V3:", consistency_score(product_v3_15), "%")
print("EXTRACTION V3:", consistency_score(extraction_v3_15), "%")
```
